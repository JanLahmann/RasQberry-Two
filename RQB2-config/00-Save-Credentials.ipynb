# Your IBM Quantum account

Every notebook runs without an account, on a simulator - start there.

To use real IBM Quantum computers, create your **own** free account at
[quantum.cloud.ibm.com](https://quantum.cloud.ibm.com) (Open Plan) and create an API key.
Then put it in the `token` line below and run that cell once; it is stored in
`~/.qiskit` for you only.
The RasQberry menu does the same: *IBM Quantum account > Save my API key*.


```python
from qiskit_ibm_runtime import QiskitRuntimeService

token = "<your-api-key>"  # the API key from quantum.cloud.ibm.com
instance = None           # optional: the CRN of one instance, e.g. "crn:v1:..."

already_saved = bool(QiskitRuntimeService.saved_accounts())
# run unedited, this cell must not replace an account saved on this Pi
if not (token or "").strip() or "<" in token or "your-api-key" in token:
    print("Your IBM Quantum account was NOT changed.")
    print("Put your API key in the token line above first - get it from https://quantum.cloud.ibm.com")
    if already_saved:
        print("An account is already saved on this Pi; saving your key here replaces it.")
else:
    QiskitRuntimeService.save_account(
        channel="ibm_quantum_platform",
        token=token,
        instance=instance,
        set_as_default=True,
        overwrite=True,
    )
    print("Saved." + (" It replaces the account saved before." if already_saved else ""))
```

Check that the saved account works (needs the internet):

```python
service = QiskitRuntimeService()
print([backend.name for backend in service.backends()])
```

Before you hand the Pi to someone else, delete your account from it.
**The cell below removes your saved API key from this Pi** - run it only when you mean to;
afterwards you need to save a key again to use IBM Quantum computers.

```python
QiskitRuntimeService.delete_account(channel="ibm_quantum_platform")
```
