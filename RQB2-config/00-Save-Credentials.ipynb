# Your IBM Quantum account

Every notebook runs without an account, on a simulator - start there.

To use real IBM Quantum computers, create your **own** free account at
[quantum.cloud.ibm.com](https://quantum.cloud.ibm.com) (Open Plan) and create an API key.
Then save it once below; it is stored in `~/.qiskit` for you only.
The RasQberry menu does the same: *IBM Quantum account > Save my API key*.


```python
from qiskit_ibm_runtime import QiskitRuntimeService

token = "<your-api-key>"  # the API key from quantum.cloud.ibm.com
instance = None           # optional: the CRN of one instance, e.g. "crn:v1:..."

QiskitRuntimeService.save_account(
    channel="ibm_quantum_platform",
    token=token,
    instance=instance,
    set_as_default=True,
    overwrite=True,
)
print("Saved.")
```

Check that the saved account works (needs the internet):

```python
service = QiskitRuntimeService()
print([backend.name for backend in service.backends()])
```

Before you hand the Pi to someone else, delete your account from it:

```python
QiskitRuntimeService.delete_account(channel="ibm_quantum_platform")
```
