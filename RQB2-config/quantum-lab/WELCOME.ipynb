# Quantum Lab

JupyterLab with Qiskit, in a container on this Pi.

**Start here:** the [IBM Quantum courses](ibm-quantum-learning/WELCOME-courses.ipynb), from IBM Quantum Learning (CC BY-SA 4.0). The folder **ibm-quantum-learning** holds them, and the tutorials in `docs/tutorials`.

**Keep your work in my-work.** That folder stays on the Pi (`~/RasQberry-Two/work/quantum-lab`). Anything saved elsewhere is gone when Quantum Lab stops.

- The course notebooks are read-only here. To change one, use *File > Save Notebook As...* and save it in `my-work/`.
- A new notebook: *File > New > Notebook*, then save it in `my-work/`.
- To stop Quantum Lab: press Enter in the window that started it (in the taskbar), or close that window.
