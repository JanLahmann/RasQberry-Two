# QuBins

JupyterLab with the Qiskit version you chose, in a container on this Pi.

**What is here** (only what you chose when it started):

- [IBM Quantum Tutorials](ibm-quantum-learning/WELCOME-tutorials.ipynb) and [IBM Quantum Courses](ibm-quantum-learning/WELCOME-courses.ipynb), from IBM Quantum Learning (CC BY-SA 4.0), in **ibm-quantum-learning**
- **quantum-paradoxes** and **fun-with-quantum**: the notebooks of these demos
- **my-work**: your own notebooks

**The notebooks are yours to change.** Your changes are kept on the Pi (`~/RasQberry-Two/work/qubins`). **originals** has the untouched versions (read-only): to restore one notebook, copy it from there. To restore all of them: *Reset notebooks to the original* in the QuBins start dialog.

**my-work** stays on the Pi too (`~/RasQberry-Two/work/quantum-lab`, shared with Qiskit Live). Anything saved elsewhere is gone when QuBins stops.

- A new notebook: *File > New > Notebook*, then save it in `my-work/`.
- To stop QuBins: press Enter in the window that started it (in the taskbar), or close that window.
