## 1. Build a circuit

A Hadamard gate (`h`) puts qubit 0 into superposition; a CNOT gate (`cx`) entangles qubit 1 with it.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
qc.draw("mpl")
```

## 2. Run it on the simulator

```python
from qiskit.primitives import StatevectorSampler

result = StatevectorSampler().run([qc], shots=1000).result()
counts = result[0].data.meas.get_counts()
counts
```

## 3. Look at the results

About half the shots give `00` and half `11`, never `01` or `10`: the two qubits always agree.

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts)
```
